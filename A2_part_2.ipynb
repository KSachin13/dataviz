```python
import sys
import copy

def read_grant_file(file_path):
    with open(file_path, 'r') as f:
        return [[int(x) for x in line.strip().split()] for line in f]

def compute_total_cost(assign, b_grant, c_grant, n):
    total = 0
    for i in range(n):
        for j in range(n):
            pet = assign[i][j]
            total += b_grant[i][j] if pet == 'B' else c_grant[i][j]

            # Neighbors (right and down only to avoid double-counting)
            if j + 1 < n and assign[i][j + 1] != pet:
                total += 1000
            if i + 1 < n and assign[i + 1][j] != pet:
                total += 1000

    return total

def greedy_initial_assignment(b_grant, c_grant, n):
    assign = [['' for _ in range(n)] for _ in range(n)]
    for i in range(n):
        for j in range(n):
            assign[i][j] = 'B' if b_grant[i][j] <= c_grant[i][j] else 'C'
    return assign

def local_search(assign, b_grant, c_grant, n, max_iter=1000):
    current = assign
    current_cost = compute_total_cost(current, b_grant, c_grant, n)

    for _ in range(max_iter):
        improved = False
        for i in range(n):
            for j in range(n):
                new_assign = copy.deepcopy(current)
                new_assign[i][j] = 'C' if current[i][j] == 'B' else 'B'
                new_cost = compute_total_cost(new_assign, b_grant, c_grant, n)
                if new_cost < current_cost:
                    current = new_assign
                    current_cost = new_cost
                    improved = True
        if not improved:
            break
    return current, current_cost

def print_assignment(assign):
    for row in assign:
        print(' '.join(row))

def main():
    if len(sys.argv) != 4:
        print("Usage: python3 label_city.py n b_grant.txt c_grant.txt")
        sys.exit(1)

    n = int(sys.argv[1])
    b_grant_file = sys.argv[2]
    c_grant_file = sys.argv[3]

    b_grant = read_grant_file(b_grant_file)
    c_grant = read_grant_file(c_grant_file)

    print("Computing optimal labeling:\n")

    initial = greedy_initial_assignment(b_grant, c_grant, n)
    best_assign, total_cost = local_search(initial, b_grant, c_grant, n)

    print_assignment(best_assign)
    print(f"\nTotal cost = {total_cost}")

if __name__ == "__main__":
    main()
```
